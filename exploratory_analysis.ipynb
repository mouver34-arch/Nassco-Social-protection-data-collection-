# Exploratory analysis — synthetic household data

> **SYNTHETIC DATA — FOR DEMONSTRATION ONLY**

This notebook is a lightweight portfolio demonstration. It does not contain real NASSCO/SOCU data.

```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt

ROOT = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
df = pd.read_csv(ROOT / 'data' / 'synthetic_household_data.csv')
df.head()
```

```python
# Basic completeness and status checks
print(df.isna().sum())
print('\nRegistration status:')
print(df['registration_status'].value_counts(dropna=False))
```

```python
# Simple synthetic visualization
df['household_size'].dropna().plot(kind='hist', bins=8)
plt.xlabel('Synthetic household size')
plt.ylabel('Number of records')
plt.title('Synthetic household-size distribution')
plt.show()
```
